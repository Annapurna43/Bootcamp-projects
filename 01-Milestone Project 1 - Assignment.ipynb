```python
def disp_board(boarddata):
    print('\t\t|\t\t|')
    print(f'\t{boarddata[0][0]}\t|\t{boarddata[0][1]}\t|\t{boarddata[0][2]}')
    print('\t\t|\t\t|')
    print('--------------------------------------------------')
    print('\t\t|\t\t|')
    print(f'\t{boarddata[1][0]}\t|\t{boarddata[1][1]}\t|\t{boarddata[1][2]}')
    print('\t\t|\t\t|')
    print('--------------------------------------------------')
    print('\t\t|\t\t|')
    print(f'\t{boarddata[2][0]}\t|\t{boarddata[2][1]}\t|\t{boarddata[2][2]}')
    print('\t\t|\t\t|')
```

```python
def player_input(message):
    s1=input(message)
    return s1
    
```

```python
def place_symbol(boarddata,sym,posi):
    match posi:
        case 1:
            boarddata[0][0]=sym
        case 2:
            boarddata[0][1]=sym
        case 3:
            boarddata[0][2]=sym
        case 4:
            boarddata[1][0]=sym
        case 5:
            boarddata[1][1]=sym
        case 6:
            boarddata[1][2]=sym
        case 7:
            boarddata[2][0]=sym
        case 8:
            boarddata[2][1]=sym
        case 9:
            boarddata[2][2]=sym
```

```python
def check_winner(boarddata,s1,s2,player1,player2):
    winner='No Winner Yet'
    win_sym=' '
    
    if (boarddata[0][0]==boarddata[1][1]==boarddata[2][2] or boarddata[0][0]==boarddata[1][0]==boarddata[2][0] or boarddata[0][0]==boarddata[0][1]==boarddata[0][2]) and boarddata[0][0]!=' ':
        win_sym=boarddata[0][0]
    elif (boarddata[1][0]==boarddata[1][1]==boarddata[1][2]) and boarddata[1][0]!=' ':
        win_sym=boarddata[1][0]
    elif (boarddata[2][0]==boarddata[2][1]==boarddata[2][2]) and  boarddata[2][0]!=' ':
        win_sym=boarddata[2][0]
    elif (boarddata[0][1]==boarddata[1][1]==boarddata[2][1]) and boarddata[0][1]!=' ':
        win_sym=boarddata[0][1]
    elif (boarddata[0][2]==boarddata[1][2]==boarddata[2][2]) and boarddata[0][2]!=' ':
        win_sym=boarddata[0][2]
    elif (boarddata[0][2]==boarddata[1][1]==boarddata[2][0]) and boarddata[0][2]!=' ':
        win_sym=boarddata[0][2]
    
    if win_sym==s1:
        winner=player1
    elif win_sym==s2:
        winner=player2
    return winner
```

```python
#new solution
from IPython.display import clear_output
boarddata=[[1,2,3],[4,5,6],[7,8,9]]
print('Below are the positions of the Tic Tac Toe Board')
disp_board(boarddata)
boarddata=[['','',''],['','',''],['','','']]
winner='No Winner Yet'
player1=player_input('Player1 Enter your name')
player2=player_input('Player2 Enter your name')
s1=player_input(f'{player1} Enter your symbol X or  O')
if s1=='X':
    s2='O'
else:
    s2='X'
positions=[100]

while winner=='No Winner Yet':
    p1=player_input(f'{player1} Enter your position')
    if p1 in positions:
        p1=player_input(f'Sorry, that position is already taken. {player1}, please Enter your position again')
    positions.append(p1)
    place_symbol(boarddata,s1,int(p1))
    winner=check_winner(boarddata,s1,s2,player1,player2)
    clear_output()
    print('After Player 1 positioning')
    disp_board(boarddata)

    if winner=='No Winner Yet':

        p2=player_input(f'{player2} Enter your position')   
        if p2 in positions:
            p2=player_input(f'Sorry, that position is already taken. {player1}, please Enter your position again')
        positions.append(p2)
        place_symbol(boarddata,s2,int(p2))
        winner=check_winner(boarddata,s1,s2,player1,player2)
        clear_output()
        print('After positioning both')
        disp_board(boarddata)
print(f'Congratulations!! The winner is {winner}')
```
